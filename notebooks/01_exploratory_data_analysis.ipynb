# Bitcoin Scam Detection Using Dynamic Graph Neural Networks
## Stage 1: Dataset Loading & Exploratory Data Analysis (EDA)

This notebook performs an interactive Exploratory Data Analysis on the **Elliptic Bitcoin Transaction Dataset**.

### Objectives:
- Load features, classes, and transaction edge list.
- Validate schema integrity and node ID matching.
- Analyze missing values, duplicate records, class distribution, and unknown vs. labelled nodes.
- Explore temporal transaction dynamics across 49 time-steps.
- Compute feature statistics (local vs aggregate features).
- Analyze network graph topology (degrees, density, subgraphs).

```python
import sys
from pathlib import Path

# Ensure project root is in sys.path
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.config import RAW_DATA_DIR, FIGURES_DIR
from src.data.loader import EllipticDataLoader
from src.data.validator import DatasetValidator
from src.eda.analyzer import EllipticEDAAnalyzer
from src.eda.visualizer import EDAVisualizer
```

## 1. Load Dataset Files

```python
loader = EllipticDataLoader(data_dir=RAW_DATA_DIR)
data_dict = loader.load_all()

print(f"Features Shape: {data_dict['features'].shape}")
print(f"Classes Shape:  {data_dict['classes'].shape}")
print(f"Edges Shape:    {data_dict['edges'].shape}")
print(f"Merged Shape:   {data_dict['merged'].shape}")
```

## 2. Dataset Schema & Validation

```python
validator = DatasetValidator(data_dict)
validation_results = validator.validate_all()
print("Validation Status:", "PASS" if validation_results['is_valid'] else "FAIL")
print("Checks Detail:")
for check_name, res in validation_results['checks'].items():
    print(f"  - {check_name}: {res}")
```

## 3. Perform Comprehensive EDA

```python
analyzer = EllipticEDAAnalyzer(data_dict)
analysis_results = analyzer.run_full_analysis()

print("=== Missing Values ===")
print(analysis_results['missing_values'])

print("\n=== Duplicate Records ===")
print(analysis_results['duplicates'])

print("\n=== Class Distribution ===")
print(analysis_results['class_distribution']['raw_counts'])
print(analysis_results['class_distribution']['percentages'])

print("\n=== Graph Statistics ===")
print(analysis_results['graph_statistics'])
```

## 4. Visualizations

```python
visualizer = EDAVisualizer(data_dict, output_dir=FIGURES_DIR)
visualizer.plot_class_distribution()
visualizer.plot_unknown_vs_labelled()
visualizer.plot_timestep_dynamics()
visualizer.plot_illicit_ratio_over_time()
visualizer.plot_node_degree_distribution()
visualizer.plot_feature_correlation()
visualizer.plot_subgraph_sample(timestep=1)
```
